```python
import pandas as pd
import numpy as np

df_raw = pd.read_csv(r"squirrels_data/squirrels.csv")
df = df_raw.copy()
```

```python
def null_percentage(df):
    return (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)

print(null_percentage(df))
```

```python
df_no_null = df.drop(columns={'Color notes', 'Other Activities', 'Other Interactions', 'Specific Location','Highlight Fur Color'})
```

```python
df_no_null.info()
```

```python
df_no_null.head()
```

```python
def clean_columns(df):
    df.columns = (
        df.columns
        .str.lower()
        .str.strip()
        .str.replace(r"\s+", "_", regex=True)
    )
    return df
df_cleaned = clean_columns(df_no_null)
```

```python
df_cleaned.info()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 8))
plt.scatter(df_cleaned["x"], df_cleaned["y"])

plt.xlabel("x")
plt.ylabel("y")
plt.title("X vs Y")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select only numeric columns
numeric_df = df_cleaned.select_dtypes(include="number")

# Create pairplot
sns.pairplot(numeric_df)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

corr = df_cleaned.corr(numeric_only=True)

plt.figure(figsize=(12, 8))
sns.heatmap(
    corr,
    annot=True,
    cmap="plasma",
    fmt=".2f",
    center=0
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
