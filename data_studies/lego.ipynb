```python
import pandas as pd
import numpy as np

colors_raw = pd.read_csv(r"lego_data/colors.csv") 
colors = colors_raw.copy()

inventories_raw = pd.read_csv(r"lego_data/inventories.csv")
inventories = inventories_raw.copy()

inventory_parts_raw = pd.read_csv(r"lego_data/inventory_parts.csv")
inventory_parts = inventory_parts_raw.copy()

part_categories_raw = pd.read_csv(r"lego_data/part_categories.csv")
part_categories = part_categories_raw.copy()

parts_raw = pd.read_csv(r"lego_data/parts.csv")
parts = parts_raw.copy()

sets_raw = pd.read_csv(r"lego_data/sets.csv")
sets = sets_raw.copy()

themes_raw = pd.read_csv(r"lego_data/themes.csv") 
themes = themes_raw.copy()
```

```python
inventory_parts_with_color = pd.merge(inventory_parts, colors, how='left', left_on='color_id', right_on='id')
inventory_parts_with_parts = pd.merge(inventory_parts_with_color, parts, how='left', left_on='part_num', right_on='part_num')
inventory_parts_with_categories = pd.merge(inventory_parts_with_parts, part_categories, how='left', left_on='part_cat_id', right_on='id')
inventory_parts_with_categories.head()
```
