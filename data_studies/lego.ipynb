```python
import pandas as pd
import numpy as np

colors_raw = pd.read_csv(r"lego_data/colors.csv") 
colors = colors_raw.copy()

inventories_raw = pd.read_csv(r"lego_data/inventories.csv")
inventories = inventories_raw.copy()

inventory_parts_raw = pd.read_csv(r"lego_data/inventory_parts.csv")
inventory_parts = inventory_parts_raw.copy()

part_categories_raw = pd.read_csv(r"lego_data/part_categories.csv")
part_categories = part_categories_raw.copy()

parts_raw = pd.read_csv(r"lego_data/parts.csv")
parts = parts_raw.copy()

sets_raw = pd.read_csv(r"lego_data/sets.csv")
sets = sets_raw.copy()

themes_raw = pd.read_csv(r"lego_data/themes.csv") 
themes = themes_raw.copy()

inventory_sets_raw = pd.read_csv(r"lego_data/inventory_sets.csv")
inventory_sets = inventory_sets_raw.copy()
```

```python
inventory_parts_with_color = pd.merge(
    inventory_parts,
    colors,
    how='left',
    left_on='color_id',
    right_on='id',
    suffixes=('', '_color')
)

inventory_parts_with_parts = pd.merge(
    inventory_parts_with_color,
    parts,
    how='left',
    left_on='part_num',
    right_on='part_num',
    suffixes=('', '_part')
)

inventory_parts_with_categories = pd.merge(
    inventory_parts_with_parts,
    part_categories,
    how='left',
    left_on='part_cat_id',
    right_on='id',
    suffixes=('', '_category')
)

inventory_parts_with_inventories = pd.merge(
    inventory_parts_with_categories,
    inventories,
    how='left',
    left_on='inventory_id',
    right_on='id',
    suffixes=('', '_inventory')
)

inventory_parts_with_sets = pd.merge(
    inventory_parts_with_inventories,
    sets,
    how='left',
    left_on='set_num',
    right_on='set_num',
    suffixes=('', '_set')
)

inventory_parts_with_themes = pd.merge(
    inventory_parts_with_sets,
    themes,
    how='left',
    left_on='theme_id',
    right_on='id',
    suffixes=('', '_theme')
)

inventory_final = pd.merge(
    inventory_parts_with_themes,
    inventory_sets,
    how='left',
    left_on='set_num',
    right_on='set_num',
    suffixes=('', '_inventory_set')
)
```
